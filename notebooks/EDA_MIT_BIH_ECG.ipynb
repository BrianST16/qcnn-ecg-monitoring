```python
# ============================================================
# EDA COMPLETO - MIT-BIH ARRHYTHMIA DATABASE
# Proyecto: CNN vs QCNN para clasificación de señales ECG
# ============================================================

# 1. INSTALAR LIBRERÍA
!pip -q install wfdb

# 2. IMPORTAR LIBRERÍAS
import wfdb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from IPython.display import display

print("=" * 70)
print("EDA - MIT-BIH ARRHYTHMIA DATABASE")
print("=" * 70)

# ============================================================
# 3. LISTA DE REGISTROS MIT-BIH
# ============================================================

registros = [
    '100','101','102','103','104','105','106','107','108','109',
    '111','112','113','114','115','116','117','118','119','121',
    '122','123','124','200','201','202','203','205','207','208',
    '209','210','212','213','214','215','217','219','220','221',
    '222','223','228','230','231','232','233','234'
]

print(f"\nCantidad de registros: {len(registros)}")

# ============================================================
# 4. CARGAR TODOS LOS REGISTROS
# ============================================================

datos = []

for nombre in registros:

    try:
        # Cargar señal
        record = wfdb.rdrecord(nombre, pn_dir='mitdb')

        # Cargar anotaciones
        annotation = wfdb.rdann(nombre, 'atr', pn_dir='mitdb')

        datos.append({
            'registro': nombre,
            'frecuencia_hz': record.fs,
            'canales': record.n_sig,
            'muestras': record.sig_len,
            'duracion_seg': record.sig_len / record.fs,
            'anotaciones': len(annotation.symbol),
            'canal_1': record.sig_name[0],
            'canal_2': record.sig_name[1] if record.n_sig > 1 else None
        })

        print(f"✓ Registro {nombre} cargado")

    except Exception as e:
        print(f"✗ Error en registro {nombre}: {e}")

# Convertir información a DataFrame
df_registros = pd.DataFrame(datos)

# ============================================================
# 5. INFORMACIÓN GENERAL
# ============================================================

print("\n")
print("=" * 70)
print("1. INFORMACIÓN GENERAL DEL DATASET")
print("=" * 70)

display(df_registros)

print("\nResumen:")
print(f"Registros analizados: {len(df_registros)}")
print(f"Frecuencia de muestreo: {df_registros['frecuencia_hz'].unique()} Hz")
print(f"Canales encontrados: {df_registros['canales'].unique()}")
print(f"Total de muestras: {df_registros['muestras'].sum():,}")
print(f"Total de anotaciones: {df_registros['anotaciones'].sum():,}")

# ============================================================
# 6. DURACIÓN DE LOS REGISTROS
# ============================================================

print("\n")
print("=" * 70)
print("2. DURACIÓN DE LOS REGISTROS")
print("=" * 70)

print(f"Duración mínima: {df_registros['duracion_seg'].min()/60:.2f} minutos")
print(f"Duración máxima: {df_registros['duracion_seg'].max()/60:.2f} minutos")
print(f"Duración promedio: {df_registros['duracion_seg'].mean()/60:.2f} minutos")

# ============================================================
# 7. CARGAR UNA SEÑAL PARA VISUALIZAR
# ============================================================

print("\n")
print("=" * 70)
print("3. VISUALIZACIÓN DE UNA SEÑAL ECG")
print("=" * 70)

record = wfdb.rdrecord('100', pn_dir='mitdb')
annotation = wfdb.rdann('100', 'atr', pn_dir='mitdb')

# Tomamos los primeros 10 segundos
segundos = 10
muestras = int(segundos * record.fs)

senal = record.p_signal[:muestras, 0]
tiempo = np.arange(muestras) / record.fs

plt.figure(figsize=(15, 5))
plt.plot(tiempo, senal)
plt.title("ECG - Registro 100 - Primeros 10 segundos")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()

# ============================================================
# 8. ANOTACIONES / LATIDOS
# ============================================================

print("\n")
print("=" * 70)
print("4. ANÁLISIS DE ANOTACIONES")
print("=" * 70)

contador = Counter(annotation.symbol)

df_anotaciones = pd.DataFrame(
    contador.items(),
    columns=['tipo', 'cantidad']
).sort_values('cantidad', ascending=False)

display(df_anotaciones)

# Gráfico
plt.figure(figsize=(12, 5))
plt.bar(df_anotaciones['tipo'], df_anotaciones['cantidad'])
plt.title("Distribución de anotaciones - Registro 100")
plt.xlabel("Tipo de anotación")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()

# ============================================================
# 9. CALIDAD DE LA SEÑAL
# ============================================================

print("\n")
print("=" * 70)
print("5. CALIDAD DE LOS DATOS")
print("=" * 70)

# Revisamos valores faltantes
valores_nan = np.isnan(record.p_signal).sum()

print(f"Valores NaN encontrados en el registro 100: {valores_nan}")

print(f"Valor mínimo de la señal: {np.min(senal):.4f}")
print(f"Valor máximo de la señal: {np.max(senal):.4f}")
print(f"Promedio de la señal: {np.mean(senal):.4f}")
print(f"Desviación estándar: {np.std(senal):.4f}")

# ============================================================
# 10. VISUALIZAR LATIDOS CON ANOTACIONES
# ============================================================

print("\n")
print("=" * 70)
print("6. ECG CON ANOTACIONES")
print("=" * 70)

# Tomar únicamente las anotaciones que aparecen
# durante los primeros 10 segundos
indices = annotation.sample
simbolos = annotation.symbol

indices_validos = indices[indices < muestras]
simbolos_validos = [
    simbolos[i]
    for i, indice in enumerate(indices)
    if indice < muestras
]

plt.figure(figsize=(15, 5))
plt.plot(tiempo, senal)

# Marcar las anotaciones
for indice, simbolo in zip(indices_validos, simbolos_validos):
    plt.scatter(
        indice / record.fs,
        senal[indice],
        s=40
    )

plt.title("ECG con anotaciones - Registro 100")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()

# ============================================================
# 11. RESUMEN FINAL DEL EDA
# ============================================================

print("\n")
print("=" * 70)
print("7. RESUMEN DEL EDA")
print("=" * 70)

print(f"""
✓ Base de datos analizada: MIT-BIH Arrhythmia Database
✓ Registros analizados: {len(df_registros)}
✓ Frecuencia de muestreo: {df_registros['frecuencia_hz'].iloc[0]} Hz
✓ Canales por registro: {df_registros['canales'].iloc[0]}
✓ Total de muestras: {df_registros['muestras'].sum():,}
✓ Total de anotaciones: {df_registros['anotaciones'].sum():,}

El EDA permite conocer:

1. La estructura general de los registros ECG.
2. La frecuencia de muestreo.
3. La duración de las señales.
4. Los tipos y cantidad de anotaciones.
5. La presencia de valores faltantes.
6. El comportamiento visual de las señales.
7. La distribución de los diferentes tipos de latidos.

IMPORTANTE:
Todavía NO estamos creando las clases NORMAL y ANORMAL.
Primero debemos definir exactamente qué anotaciones de MIT-BIH
vamos a considerar normales y cuáles anormales para nuestro proyecto.
""")

print("=" * 70)
print("EDA TERMINADO")
print("=" * 70)
```

```python
# ============================================================
# PREPARACIÓN DEL DATASET ECG
# Proyecto: CNN vs QCNN - MIT-BIH Arrhythmia Database
# ============================================================

# ============================================================
# 1. INSTALAR LIBRERÍAS NECESARIAS
# ============================================================

!pip -q install scipy scikit-learn

# ============================================================
# 2. IMPORTAR LIBRERÍAS
# ============================================================

import wfdb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import butter, filtfilt
from sklearn.model_selection import train_test_split
from collections import Counter

print("=" * 70)
print("PREPARACIÓN DEL DATASET ECG")
print("=" * 70)


# ============================================================
# 3. REGISTROS MIT-BIH
# ============================================================

registros = [
    '100','101','102','103','104','105','106','107','108','109',
    '111','112','113','114','115','116','117','118','119','121',
    '122','123','124','200','201','202','203','205','207','208',
    '209','210','212','213','214','215','217','219','220','221',
    '222','223','228','230','231','232','233','234'
]

print(f"\nRegistros disponibles: {len(registros)}")


# ============================================================
# 4. DEFINICIÓN DE LAS CLASES
# ============================================================

# N = latido normal
CLASES_NORMALES = [
    'N'
]

# Latidos considerados anormales para este proyecto
CLASES_ANORMALES = [
    'L',   # Left bundle branch block beat
    'R',   # Right bundle branch block beat
    'A',   # Atrial premature beat
    'V',   # Premature ventricular contraction
    'F',   # Fusion of ventricular and normal beat
    '/',   # Paced beat
    'f',   # Fusion of paced and normal beat
    'j',   # Nodal (junctional) escape beat
    'a',   # Aberrated atrial premature beat
    'E',   # Ventricular escape beat
    'J',   # Nodal (junctional) premature beat
    'S'    # Supraventricular premature beat
]

print("\nCLASES DEFINIDAS")
print("----------------")
print("NORMAL :", CLASES_NORMALES)
print("ANORMAL:", CLASES_ANORMALES)


# ============================================================
# 5. PARÁMETROS DEL PREPROCESAMIENTO
# ============================================================

FRECUENCIA_MUESTREO = 360

# Filtro pasa-banda
FRECUENCIA_BAJA = 0.5
FRECUENCIA_ALTA = 40

# Ventana alrededor del pico R
# 0.8 segundos = 288 muestras a 360 Hz
DURACION_VENTANA = 0.8

# 0.4 segundos antes y 0.4 después del pico R
MUESTRAS_ANTES = int(0.4 * FRECUENCIA_MUESTREO)
MUESTRAS_DESPUES = int(0.4 * FRECUENCIA_MUESTREO)

LONGITUD_LATIDO = MUESTRAS_ANTES + MUESTRAS_DESPUES

print("\nPARÁMETROS")
print("----------")
print(f"Frecuencia de muestreo: {FRECUENCIA_MUESTREO} Hz")
print(f"Filtro: {FRECUENCIA_BAJA} - {FRECUENCIA_ALTA} Hz")
print(f"Ventana por latido: {DURACION_VENTANA} segundos")
print(f"Muestras por latido: {LONGITUD_LATIDO}")


# ============================================================
# 6. FUNCIÓN PARA FILTRAR LA SEÑAL
# ============================================================

def filtrar_ecg(senal, fs=360, baja=0.5, alta=40):

    nyquist = fs / 2

    frecuencia_baja = baja / nyquist
    frecuencia_alta = alta / nyquist

    b, a = butter(
        4,
        [frecuencia_baja, frecuencia_alta],
        btype='band'
    )

    senal_filtrada = filtfilt(b, a, senal)

    return senal_filtrada


# ============================================================
# 7. FUNCIÓN PARA NORMALIZAR LA SEÑAL
# ============================================================

def normalizar_senal(senal):

    media = np.mean(senal)
    desviacion = np.std(senal)

    if desviacion == 0:
        return senal

    return (senal - media) / desviacion


# ============================================================
# 8. CREAR DATASET
# ============================================================

X = []
y = []
registros_latidos = []
simbolos_latidos = []

print("\n")
print("=" * 70)
print("PROCESANDO REGISTROS")
print("=" * 70)

for nombre in registros:

    try:

        # ----------------------------------------------------
        # Cargar señal
        # ----------------------------------------------------

        record = wfdb.rdrecord(
            nombre,
            pn_dir='mitdb'
        )

        # ----------------------------------------------------
        # Cargar anotaciones
        # ----------------------------------------------------

        annotation = wfdb.rdann(
            nombre,
            'atr',
            pn_dir='mitdb'
        )

        # ----------------------------------------------------
        # Tomar primer canal
        # ----------------------------------------------------

        senal = record.p_signal[:, 0]

        # ----------------------------------------------------
        # Filtrar
        # ----------------------------------------------------

        senal_filtrada = filtrar_ecg(
            senal,
            fs=record.fs,
            baja=FRECUENCIA_BAJA,
            alta=FRECUENCIA_ALTA
        )

        # ----------------------------------------------------
        # Normalizar
        # ----------------------------------------------------

        senal_normalizada = normalizar_senal(
            senal_filtrada
        )

        contador_registro = 0

        # ----------------------------------------------------
        # Recorrer anotaciones
        # ----------------------------------------------------

        for posicion, simbolo in zip(
            annotation.sample,
            annotation.symbol
        ):

            # ------------------------------------------------
            # Determinar clase
            # ------------------------------------------------

            if simbolo in CLASES_NORMALES:

                etiqueta = 0

            elif simbolo in CLASES_ANORMALES:

                etiqueta = 1

            else:

                # Ignorar anotaciones que no pertenecen
                # a ninguna de nuestras dos clases
                continue

            # ------------------------------------------------
            # Comprobar límites de la ventana
            # ------------------------------------------------

            inicio = posicion - MUESTRAS_ANTES
            fin = posicion + MUESTRAS_DESPUES

            if inicio < 0 or fin > len(senal_normalizada):
                continue

            # ------------------------------------------------
            # Extraer latido
            # ------------------------------------------------

            latido = senal_normalizada[inicio:fin]

            # Verificar tamaño
            if len(latido) != LONGITUD_LATIDO:
                continue

            # ------------------------------------------------
            # Guardar
            # ------------------------------------------------

            X.append(latido)
            y.append(etiqueta)

            registros_latidos.append(nombre)
            simbolos_latidos.append(simbolo)

            contador_registro += 1

        print(
            f"✓ Registro {nombre}: "
            f"{contador_registro} latidos válidos"
        )

    except Exception as e:

        print(
            f"✗ Error en registro {nombre}: {e}"
        )


# ============================================================
# 9. CONVERTIR A NUMPY
# ============================================================

X = np.array(X, dtype=np.float32)
y = np.array(y, dtype=np.int64)

registros_latidos = np.array(registros_latidos)
simbolos_latidos = np.array(simbolos_latidos)

print("\n")
print("=" * 70)
print("DATASET CREADO")
print("=" * 70)

print(f"Forma de X: {X.shape}")
print(f"Forma de y: {y.shape}")

print(f"\nTotal de latidos: {len(X)}")


# ============================================================
# 10. DISTRIBUCIÓN DE LAS CLASES
# ============================================================

cantidad_normal = np.sum(y == 0)
cantidad_anormal = np.sum(y == 1)

print("\n")
print("=" * 70)
print("DISTRIBUCIÓN DE CLASES")
print("=" * 70)

print(f"Latidos NORMALES : {cantidad_normal:,}")
print(f"Latidos ANORMALES: {cantidad_anormal:,}")

porcentaje_normal = cantidad_normal / len(y) * 100
porcentaje_anormal = cantidad_anormal / len(y) * 100

print(f"\nPorcentaje NORMAL : {porcentaje_normal:.2f}%")
print(f"Porcentaje ANORMAL: {porcentaje_anormal:.2f}%")


# ============================================================
# 11. GRÁFICO DE BALANCE
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    ['NORMAL', 'ANORMAL'],
    [cantidad_normal, cantidad_anormal]
)

plt.title("Distribución de clases del dataset")
plt.xlabel("Clase")
plt.ylabel("Cantidad de latidos")
plt.grid(axis='y')

plt.show()


# ============================================================
# 12. DISTRIBUCIÓN DE SÍMBOLOS
# ============================================================

contador_simbolos = Counter(simbolos_latidos)

df_simbolos = pd.DataFrame(
    contador_simbolos.items(),
    columns=['simbolo', 'cantidad']
).sort_values(
    'cantidad',
    ascending=False
)

print("\n")
print("=" * 70)
print("DISTRIBUCIÓN DE ANOTACIONES")
print("=" * 70)

display(df_simbolos)


# ============================================================
# 13. VISUALIZAR LATIDOS NORMALES Y ANORMALES
# ============================================================

indice_normal = np.where(y == 0)[0]
indice_anormal = np.where(y == 1)[0]

plt.figure(figsize=(15, 5))

if len(indice_normal) > 0:
    plt.plot(
        X[indice_normal[0]],
        label='Normal'
    )

if len(indice_anormal) > 0:
    plt.plot(
        X[indice_anormal[0]],
        label='Anormal'
    )

plt.title("Ejemplo de latido normal y anormal")
plt.xlabel("Muestras")
plt.ylabel("Amplitud normalizada")
plt.legend()
plt.grid()

plt.show()


# ============================================================
# 14. VERIFICAR VALORES NaN
# ============================================================

nan_X = np.isnan(X).sum()
nan_y = np.isnan(y).sum()

print("\n")
print("=" * 70)
print("CONTROL DE CALIDAD")
print("=" * 70)

print(f"Valores NaN en X: {nan_X}")
print(f"Valores NaN en y: {nan_y}")

if nan_X == 0 and nan_y == 0:
    print("✓ No existen valores NaN")
else:
    print("⚠ Existen valores NaN")


# ============================================================
# 15. CREAR TABLA DEL DATASET
# ============================================================

df_dataset = pd.DataFrame({
    'registro': registros_latidos,
    'simbolo': simbolos_latidos,
    'clase': y
})

df_dataset['nombre_clase'] = df_dataset['clase'].map({
    0: 'NORMAL',
    1: 'ANORMAL'
})

print("\n")
print("=" * 70)
print("TABLA DEL DATASET")
print("=" * 70)

display(df_dataset.head(10))


# ============================================================
# 16. PARTICIÓN POR REGISTRO
# ============================================================
#
# IMPORTANTE:
# No debemos mezclar latidos del mismo registro entre
# entrenamiento, validación y prueba.
#
# De esta manera evitamos que el modelo vea durante el
# entrenamiento latidos del mismo paciente que aparecen
# posteriormente en prueba.
# ============================================================

registros_unicos = np.unique(registros_latidos)

print("\n")
print("=" * 70)
print("PARTICIÓN POR REGISTRO")
print("=" * 70)

print(f"Registros disponibles para división: {len(registros_unicos)}")


# ============================================================
# 17. DIVIDIR REGISTROS
# ============================================================

registros_train, registros_test = train_test_split(
    registros_unicos,
    test_size=0.20,
    random_state=42
)

registros_train, registros_validacion = train_test_split(
    registros_train,
    test_size=0.20,
    random_state=42
)

print(f"Registros entrenamiento : {len(registros_train)}")
print(f"Registros validación    : {len(registros_validacion)}")
print(f"Registros prueba        : {len(registros_test)}")


# ============================================================
# 18. CREAR MÁSCARAS
# ============================================================

mascara_train = np.isin(
    registros_latidos,
    registros_train
)

mascara_validacion = np.isin(
    registros_latidos,
    registros_validacion
)

mascara_test = np.isin(
    registros_latidos,
    registros_test
)


# ============================================================
# 19. CREAR CONJUNTOS
# ============================================================

X_train = X[mascara_train]
y_train = y[mascara_train]

X_validacion = X[mascara_validacion]
y_validacion = y[mascara_validacion]

X_test = X[mascara_test]
y_test = y[mascara_test]


# ============================================================
# 20. INFORMACIÓN DE LOS CONJUNTOS
# ============================================================

print("\n")
print("=" * 70)
print("CONJUNTOS FINALES")
print("=" * 70)

print("\nENTRENAMIENTO")
print(f"X_train: {X_train.shape}")
print(f"y_train: {y_train.shape}")

print("\nVALIDACIÓN")
print(f"X_validacion: {X_validacion.shape}")
print(f"y_validacion: {y_validacion.shape}")

print("\nPRUEBA")
print(f"X_test: {X_test.shape}")
print(f"y_test: {y_test.shape}")


# ============================================================
# 21. DISTRIBUCIÓN POR CONJUNTO
# ============================================================

print("\n")
print("=" * 70)
print("DISTRIBUCIÓN POR CONJUNTO")
print("=" * 70)

print("\nENTRENAMIENTO")
print(
    f"Normal: {np.sum(y_train == 0):,} | "
    f"Anormal: {np.sum(y_train == 1):,}"
)

print("\nVALIDACIÓN")
print(
    f"Normal: {np.sum(y_validacion == 0):,} | "
    f"Anormal: {np.sum(y_validacion == 1):,}"
)

print("\nPRUEBA")
print(
    f"Normal: {np.sum(y_test == 0):,} | "
    f"Anormal: {np.sum(y_test == 1):,}"
)


# ============================================================
# 22. GUARDAR DATASET
# ============================================================

np.save(
    'X_train.npy',
    X_train
)

np.save(
    'y_train.npy',
    y_train
)

np.save(
    'X_validacion.npy',
    X_validacion
)

np.save(
    'y_validacion.npy',
    y_validacion
)

np.save(
    'X_test.npy',
    X_test
)

np.save(
    'y_test.npy',
    y_test
)

df_dataset.to_csv(
    'dataset_ecg.csv',
    index=False
)

print("\n")
print("=" * 70)
print("ARCHIVOS GUARDADOS")
print("=" * 70)

print("""
✓ X_train.npy
✓ y_train.npy
✓ X_validacion.npy
✓ y_validacion.npy
✓ X_test.npy
✓ y_test.npy
✓ dataset_ecg.csv
""")


# ============================================================
# 23. RESUMEN FINAL
# ============================================================

print("=" * 70)
print("RESUMEN DE PREPARACIÓN")
print("=" * 70)

print(f"""
Base de datos       : MIT-BIH Arrhythmia Database

Registros utilizados: {len(registros)}

Clases:
  0 → NORMAL
  1 → ANORMAL

Frecuencia          : {FRECUENCIA_MUESTREO} Hz
Filtro              : {FRECUENCIA_BAJA}-{FRECUENCIA_ALTA} Hz
Ventana             : {DURACION_VENTANA} segundos
Muestras por latido : {LONGITUD_LATIDO}

Total de latidos    : {len(X):,}

Entrenamiento       : {len(X_train):,}
Validación          : {len(X_validacion):,}
Prueba              : {len(X_test):,}

✓ Filtrado aplicado
✓ Normalización aplicada
✓ Segmentación por latido
✓ Clases binarias creadas
✓ Partición por registro
✓ Dataset guardado

El dataset está listo para la siguiente etapa:
ENTRENAMIENTO DE LA CNN 1-D
""")

print("=" * 70)
print("PREPARACIÓN DEL DATASET TERMINADA")
print("=" * 70)
```

```python
# ============================================================
# CNN 1-D PARA CLASIFICACIÓN DE SEÑALES ECG
# NORMAL (0) vs ANORMAL (1)
# ============================================================

# 1. INSTALAR TENSORFLOW
!pip -q install tensorflow scikit-learn

# 2. IMPORTAR LIBRERÍAS
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

# ============================================================
# 3. REVISAR LOS DATOS
# ============================================================

print("Datos de entrenamiento:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nDatos de validación:")
print("X_validacion:", X_validacion.shape)
print("y_validacion:", y_validacion.shape)

print("\nDatos de prueba:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# ============================================================
# 4. PREPARAR LA ENTRADA PARA CNN 1-D
# ============================================================
# La CNN necesita:
# (cantidad de ejemplos, número de muestras, canales)

if len(X_train.shape) == 2:
    X_train_cnn = X_train[..., np.newaxis]
    X_validacion_cnn = X_validacion[..., np.newaxis]
    X_test_cnn = X_test[..., np.newaxis]
else:
    X_train_cnn = X_train
    X_validacion_cnn = X_validacion
    X_test_cnn = X_test

print("\nForma final para CNN:")
print("X_train:", X_train_cnn.shape)
print("X_validacion:", X_validacion_cnn.shape)
print("X_test:", X_test_cnn.shape)

# ============================================================
# 5. CREAR LA CNN 1-D
# ============================================================

modelo_cnn = Sequential([

    # Primera capa convolucional
    Conv1D(
        filters=32,
        kernel_size=5,
        activation='relu',
        input_shape=(X_train_cnn.shape[1], 1)
    ),

    # Reduce el tamaño de la señal
    MaxPooling1D(pool_size=2),

    # Segunda capa convolucional
    Conv1D(
        filters=64,
        kernel_size=5,
        activation='relu'
    ),

    # Segundo pooling
    MaxPooling1D(pool_size=2),

    # Convierte las características en un vector
    Flatten(),

    # Capa completamente conectada
    Dense(64, activation='relu'),

    # Evita sobreajuste
    Dropout(0.3),

    # Salida binaria
    Dense(1, activation='sigmoid')
])

# ============================================================
# 6. MOSTRAR LA ARQUITECTURA
# ============================================================

modelo_cnn.summary()

# ============================================================
# 7. COMPILAR EL MODELO
# ============================================================

modelo_cnn.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ============================================================
# 8. ENTRENAMIENTO
# ============================================================

# Detiene el entrenamiento si la validación deja de mejorar
detener = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

historial = modelo_cnn.fit(
    X_train_cnn,
    y_train,
    validation_data=(X_validacion_cnn, y_validacion),
    epochs=30,
    batch_size=64,
    callbacks=[detener],
    verbose=1
)

# ============================================================
# 9. GRAFICAR EL ENTRENAMIENTO
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(historial.history['accuracy'], label='Entrenamiento')
plt.plot(historial.history['val_accuracy'], label='Validación')

plt.title('Accuracy durante el entrenamiento')
plt.xlabel('Época')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 5))

plt.plot(historial.history['loss'], label='Entrenamiento')
plt.plot(historial.history['val_loss'], label='Validación')

plt.title('Pérdida durante el entrenamiento')
plt.xlabel('Época')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()

# ============================================================
# 10. EVALUAR EN EL CONJUNTO DE PRUEBA
# ============================================================

resultado = modelo_cnn.evaluate(
    X_test_cnn,
    y_test,
    verbose=0
)

print("\nResultado del conjunto de prueba:")
print("Loss:", resultado[0])
print("Accuracy:", resultado[1])

# ============================================================
# 11. HACER PREDICCIONES
# ============================================================

probabilidades = modelo_cnn.predict(
    X_test_cnn,
    verbose=0
).ravel()

# Si probabilidad >= 0.5 -> ANORMAL (1)
# Si probabilidad < 0.5 -> NORMAL (0)

y_pred = (probabilidades >= 0.5).astype(int)

# ============================================================
# 12. MÉTRICAS
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

sensibilidad = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

print("\n==============================")
print("RESULTADOS CNN 1-D")
print("==============================")

print(f"Accuracy:     {accuracy:.4f}")
print(f"Precision:    {precision:.4f}")
print(f"Sensibilidad: {sensibilidad:.4f}")
print(f"F1-score:     {f1:.4f}")

# ============================================================
# 13. REPORTE COMPLETO
# ============================================================

print("\nReporte de clasificación:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=['NORMAL', 'ANORMAL'],
        zero_division=0
    )
)

# ============================================================
# 14. MATRIZ DE CONFUSIÓN
# ============================================================

matriz = confusion_matrix(y_test, y_pred)

print("\nMatriz de confusión:")
print(matriz)

disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=['NORMAL', 'ANORMAL']
)

disp.plot()
plt.title("Matriz de confusión - CNN 1-D")
plt.show()

# ============================================================
# 15. EJEMPLOS DE PREDICCIONES
# ============================================================

print("\nPrimeras 10 predicciones:")

for i in range(min(10, len(y_test))):

    clase_real = "NORMAL" if y_test[i] == 0 else "ANORMAL"
    clase_predicha = "NORMAL" if y_pred[i] == 0 else "ANORMAL"

    print(
        f"Ejemplo {i+1}: "
        f"Real = {clase_real} | "
        f"Predicción = {clase_predicha} | "
        f"Probabilidad anormal = {probabilidades[i]:.3f}"
    )

# ============================================================
# 16. GUARDAR EL MODELO
# ============================================================

modelo_cnn.save("modelo_cnn_ecg.keras")

print("\nModelo guardado como: modelo_cnn_ecg.keras")

# ============================================================
# 17. GUARDAR RESULTADOS
# ============================================================

resultados_cnn = {
    "accuracy": accuracy,
    "precision": precision,
    "sensibilidad": sensibilidad,
    "f1": f1
}

np.save(
    "resultados_cnn.npy",
    resultados_cnn,
    allow_pickle=True
)

print("Resultados guardados.")

# ============================================================
# FIN DE LA CNN 1-D
# ============================================================

print("\n==========================================")
print("CNN 1-D COMPLETADA")
print("==========================================")
print("Ya tenemos nuestro modelo clásico de referencia.")
print("El siguiente paso será preparar la información")
print("para construir la QCNN.")
```
